```python
import pandas as pd
import numpy as np
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
def wait_for_page_to_load(driver, wait):
    title = driver.title
    #within given time 5sec if all page get load i.e complete then good to proceed
    try:
        wait.until(
            lambda d: d.execute_script("return document.readyState") == "complete"
        )
    except:
        print(f"The webpage \"{title}\" did not get fully loaded.\n")
    #if webpage load completely in given duration
    else:
        print(f"The webpage \"{title}\" did get fully loaded.\n")
```

# 1.Base code

```python
driver = webdriver.Chrome()
driver.maximize_window()

#explicit wait
wait = WebDriverWait(driver, 5)

#accessing the target webpage
url = "https://www.magicbricks.com/"
driver.get(url)
wait_for_page_to_load(driver, wait)



#identifying search bar 
try:
    search_bar = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="keyword"]'))
    )
except:
    print("Timeout while locating Search Bar.\n")
else:
    #search and cancel mumbai tag
    try:
        search_bar.click()
        clk = wait.until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="keyword_autoSuggestSelectedDiv"]/div/div[2]'))
        )
        clk.click()
    except:
        print("mumbai tag not found")
    finally: #enter mumbai in search bar
        search_bar.click()
        search_bar.send_keys("mumbai")
        time.sleep(5)



#selecting valid option from list
try:
    valid_option = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="serachSuggest"]/div[3]'))
    )
except:
    print("Timeout while locating valid search option.\n")
else:
    valid_option.click()
    time.sleep(2)



#click on search button
try:
    search_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="searchFormHolderSection"]/section/div/div[1]/div[3]/div[4]'))
    )
except:
    print("Timeout while clicking on \"search\" button.\n")
else:
    search_button.click()
    wait_for_page_to_load(driver, wait)



# click on property type button
try:
    property_type_button = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[3]/div/div[1]'))
    )
except:
    print("Timeout while locating \"property_type_button\" button.\n")
else:
    driver.execute_script(
        "window.scrollBy(0, arguments[0].getBoundingClientRect().top - 100);",property_type_button
    )
    time.sleep(2)

    # find the element again because DOM may have changed
    property_type_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[3]/div/div[1]'))
    )

    property_type_button.click()
    time.sleep(2)



# unchecked House/Villa
try:
    house_villa = wait.until(
        EC.presence_of_element_located((By.ID, "propResidential_propertyType_10001_10017"))
    )
except:
    print("Timeout while locating \"House/Villa\" checkbox.\n")
else:
    if house_villa.is_selected():
        house_villa_label = wait.until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[3]/div/div[2]/div[1]/div[2]/div[2]/label'))
        )
        house_villa_label.click()
        time.sleep(2)

        

#clicking on done
try:
    click_done = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[3]/div/div[2]/div[4]'))
    )
except:
    print("Timeout while locating \"done\".\n")
else:
    click_done.click()
    time.sleep(2)



# click on BHK button
try:
    BHK_button = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[4]/div/div[1]/div'))
    )
except:
    print("Timeout while locating \"BHK\" button.\n")
else:
    driver.execute_script(
        "window.scrollBy(0, arguments[0].getBoundingClientRect().top - 100);",BHK_button
    )
    time.sleep(2)

    # find the element again because DOM may have changed
    BHK_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[4]/div/div[1]/div'))
    )

    BHK_button.click()
    time.sleep(2)



# unchecked 2BHK
try:
    BHK2 = wait.until(
        EC.presence_of_element_located((By.ID, "bhk_11701"))
    )
except:
    print("Timeout while locating \"2 BHK\" checkbox.\n")
else:
    if BHK2.is_selected():
        BHK2_label = wait.until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[4]/div/div[2]/div[1]/div[2]/div[2]/label'))
        )
        BHK2_label.click()
        time.sleep(2)

        

# unchecked 3BHK
try:
    BHK3 = wait.until(
        EC.presence_of_element_located((By.ID, "bhk_11702"))
    )
except:
    print("Timeout while locating \"3 BHK\" checkbox.\n")
else:
    if BHK3.is_selected():
        BHK3_label = wait.until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[4]/div/div[2]/div[1]/div[2]/div[3]/label'))
        )
        BHK3_label.click()
        time.sleep(2)

        

#clicking on done
try:
    click_done = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[4]/div/div[2]/div[2]'))
    )
except:
    print("Timeout while locating \"done\".\n")
else:
    click_done.click()
    time.sleep(2)


# click on More Filters button
try:
    more_filters_button = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[6]/div/div[1]'))
    )
except:
    print("Timeout while locating \"more_filters\" button.\n")
else:
    driver.execute_script(
        "window.scrollBy(0, arguments[0].getBoundingClientRect().top - 100);",more_filters_button
    )
    time.sleep(2)

    # find the element again because DOM may have changed
    more_filters_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="body"]/div[1]/div/div[2]/div[6]/div/div[1]'))
    )

    more_filters_button.click()
    time.sleep(2)



# checked ready to move button
try:
    ready_to_move_button = wait.until(
        EC.presence_of_element_located((By.ID, "possessionStatus_10081"))
    )
except:
    print("Timeout while locating \"ready_to_move\" button.\n")
else:
    ready_to_move_label = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreFilter_1"]/div[2]/div[1]/label'))
    )
    ready_to_move_label.click()
    time.sleep(2)

        
# checked verified properties button
try:
    verified_properties_button = wait.until(
        EC.presence_of_element_located((By.ID, "verifiedProperties_Y"))
    )
except:
    print("Timeout while locating \"verified_properties\" button.\n")
else:
    verified_properties_label = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreFilter_11"]/div[2]/div/label'))
    )
    verified_properties_label.click()
    time.sleep(2)


# checked photos button
try:
    photos_button = wait.until(
        EC.presence_of_element_located((By.ID, "imageVideo_1"))
    )
except:
    print("Timeout while locating \"photos\" button.\n")
else:
    photos_label = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreFilter_12"]/div[2]/div[1]/label'))
    )
    photos_label.click()
    time.sleep(2)



# checked videos button
try:
    videos_button = wait.until(
        EC.presence_of_element_located((By.ID, "imageVideo_2"))
    )
except:
    print("Timeout while locating \"videos\" button.\n")
else:
    videos_label = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreFilter_12"]/div[2]/div[2]/label'))
    )
    videos_label.click()
    time.sleep(2)


# checked rera properties button
try:
    rera_properties_button = wait.until(
        EC.presence_of_element_located((By.ID, "reraProperties_Y"))
    )
except:
    print("Timeout while locating \"rera_properties\" button.\n")
else:
    rera_properties_label = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreFilter_20"]/div[2]/div/label'))
    )
    rera_properties_label.click()
    time.sleep(2)


#clicking on view selected properties
try:
    view_selected_properties_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="moreRefined"]/div[3]/div/div[2]'))
    )
except:
    print("Timeout while locating \"view_selected_properties\" button.\n")
else:
    view_selected_properties_button.click()
    time.sleep(3)

    
property_data = []
processed_count = 0

height = driver.execute_script("return document.body.scrollHeight")
while True:
    # scroll to bottom
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight)"
    )

    time.sleep(3)

    # find all property cards
    property_cards = driver.find_elements(By.CLASS_NAME, "mb-srp__list")

    print("Total cards:", len(property_cards))

    # Take only newly loaded cards
    new_cards = property_cards[processed_count:]

    print("New cards:", len(new_cards))
    
    for card in new_cards:

        # clicking on down arrrow in each property
        try:
            down_arrow_button = card.find_element(
                By.CSS_SELECTOR,"div.mb-srp__card__summary__action"
            )
        
            driver.execute_script(
                "arguments[0].scrollIntoView({block: 'center'});",down_arrow_button
            )
        
            time.sleep(0.5)
        
            driver.execute_script(
                "arguments[0].click();",down_arrow_button
            )
        
            time.sleep(1)
        
        except Exception as e:
            print("Down arrow not found")

        
        # property id
        try:
            property_id = card.get_attribute("id").strip()
            print(property_id)
        except:
            property_id = np.nan
    
        # property title
        try:
            property_title = card.find_element(By.CLASS_NAME, "mb-srp__card--title").text.strip()
            #print(property_title)
        except:
            property_title = np.nan
            
    
        # property developer name
        try:
            property_developer = card.find_element(By.CLASS_NAME, "mb-srp__card__developer--name--highlight").text.strip()
            #print(property_developer)
        except:
            property_developer = np.nan
            
    
        # carpet area
        try:
            property_carpet_area = card.find_element(
                By.XPATH,".//div[text()='Carpet Area']/following-sibling::div"
            ).text.strip()
            #print(property_carpet_area)
        except:
            property_carpet_area = np.nan
            
    
        # status 
        try:
            property_status = card.find_element(
                By.XPATH,".//div[text()='Status']/following-sibling::div"
            ).text.strip()
            #print(property_status)
        except:
            property_status = np.nan
            
    
        # Transaction 
        try:
            property_transaction = card.find_element(
                By.XPATH,".//div[text()='Transaction']/following-sibling::div"
            ).text.strip()
            #print(property_transaction)
        except:
            property_transaction = np.nan


        # Furnishing 
        try:
            property_furnishing = card.find_element(
                By.XPATH,".//div[text()='Furnishing']/following-sibling::div"
            ).text.strip()
            #print(property_furnishing)
        except:
            property_furnishing = np.nan


        # Society 
        try:
            property_society = card.find_element(
                By.XPATH,".//div[text()='Society']/following-sibling::div"
            ).text.strip()
            #print(property_society)
        except:
            property_society = np.nan


        # Car Parking 
        try:
            property_car_parking = card.find_element(
                By.XPATH,".//div[text()='Car Parking']/following-sibling::div"
            ).text.strip()
            #print(property_car_parking)
        except:
            property_car_parking = np.nan


        # Floor 
        try:
            property_floor = card.find_element(
                By.XPATH,".//div[text()='Floor']/following-sibling::div"
            ).text.strip()
            #print(property_floor)
        except:
            property_floor = np.nan


        # facing 
        try:
            property_facing = card.find_element(
                By.XPATH,".//div[text()='facing']/following-sibling::div"
            ).text.strip()
            #print(property_facing)
        except:
            property_facing = np.nan


        # overlooking 
        try:
            property_overlooking = card.find_element(
                By.XPATH,".//div[text()='overlooking']/following-sibling::div"
            ).text.strip()
            #print(property_overlooking)
        except:
            property_overlooking = np.nan


        # Super Area 
        try:
            property_super_area = card.find_element(
                By.XPATH,".//div[text()='Super Area']/following-sibling::div"
            ).text.strip()
            #print(property_super_area)
        except:
            property_super_area = np.nan


        # Price
        try:
            property_price = card.find_element(
                By.CLASS_NAME,"mb-srp__card__price--amount"
            ).text.strip()
            #print(property_price)
        except:
            property_price = np.nan
        
        # Cost per sqft
        try:
            property_cost_per_sqft = card.find_element(
                By.CLASS_NAME,"mb-srp__card__price--size"
            ).text.strip()
            #print(property_cost_per_sqft)
        except:
            property_cost_per_sqft = np.nan


        all_features = {
            "property_id" : property_id,
            "property_cost_per_sqft" : property_cost_per_sqft,
            "property_price" : property_price,
            "property_super_area" : property_super_area,
            "property_overlooking" : property_overlooking,
            "property_facing" : property_facing,
            "property_floor" : property_floor,
            "property_car_parking" : property_car_parking,
            "property_society" : property_society,
            "property_furnishing" : property_furnishing,
            "property_transaction" : property_transaction,
            "property_status" : property_status,
            "property_carpet_area" : property_carpet_area,
            "property_developer" : property_developer,
            "property_title" : property_title
        }

        property_data.append(all_features)


    # Update processed count
    processed_count = len(property_cards)

    # NOW check the page height
    new_height = driver.execute_script(
        "return document.body.scrollHeight"
    )

    print("Old height:", height)
    print("New height:", new_height)

    # If page didn't grow, stop
    if height == new_height:
        break

    # Update height for next iteration
    height = new_height

time.sleep(2)
driver.quit()


df = pd.DataFrame(property_data)
df.to_csv("property_data.csv" , index = False, encoding="utf-8-sig")
```
